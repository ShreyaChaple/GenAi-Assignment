```python
import re
import inflect
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
p = inflect.engine()
```

```python
stop_words = set(stopwords.words('english'))
```

```python
def filter_non_alphanumeric(word):
    return re.sub(r'[^a-zA-Z0-9]', '', word)
```

```python
def convert_numbers_to_words(word):
    if word.isdigit():  # Check if word is a number
        return p.number_to_words(word)
    return word
```

```python
def is_valid_word(word):
    return word and word.isalnum()
```

```python
def text_processing_pipeline(input_text):
    print(f"Original Text: {input_text}\n")

    # Step 1: Split text into words
    words = word_tokenize(input_text)
    print(f"Words After Splitting: {words}\n")

    processed_words = []

    for word in words:
        # Step 2: Filter out non-alphanumeric characters
        filtered_word = filter_non_alphanumeric(word)

        # Step 3: Remove stop words
        if filtered_word.lower() in stop_words:
            continue

        # Step 4: Handle numbers (convert to words)
        if re.search(r'\d', filtered_word):  # If word contains digits
            filtered_word = convert_numbers_to_words(filtered_word)

        # Step 5: Store valid words
        if is_valid_word(filtered_word):
            processed_words.append(filtered_word)

    print(f"Processed Words: {processed_words}")

    # Final step: Store parsed data (for this example, we will just return it)
    return processed_words
```

```python
input_text = "Hello! I have 2 apples, 3 oranges, and a cat. They cost $10 in total!"
```

```python
processed_text = text_processing_pipeline(input_text)
```
